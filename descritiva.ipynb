## 2. Medidas de Dispersão
Saber a média não é suficiente. Precisamos saber o quanto os dados variam em torno dessa média. Isso nos diz se nossos dados são consistentes ou muito espalhados.

### Exemplo: Desvio Padrão (Standard Deviation)
O desvio padrão indica o quanto, em média, os valores se afastam da média. Um desvio padrão alto significa dados muito espalhados.

```python
# [EXEMPLO] Calculando o desvio padrão da idade
desvio_padrao_idade = df['age'].std()
print(f"O desvio padrão da idade é de {desvio_padrao_idade:.2f} anos.")
```

# 🧹 Limpeza e Preparação de Dados para I.A.

Modelos de Inteligência Artificial são como motores de alta performance: eles precisam de combustível limpo para funcionar bem. A maioria dos algoritmos (como os do `scikit-learn`) não consegue processar "buracos" nos dados (valores nulos) ou textos (dados categóricos). 

Nesta etapa, vamos aprender a tratar dados ausentes e a transformar texto em números.

```python
# [EXEMPLO] Identificando Valores Nulos
# O método .isnull() verifica se o dado está vazio, e o .sum() soma essas ocorrências por coluna.

valores_nulos = df.isnull().sum()
print("Quantidade de dados faltando em cada coluna:")
print(valores_nulos[valores_nulos > 0]) # Mostra apenas as colunas com problemas
```

---
## 1. Tratando Dados Ausentes (Imputação e Exclusão)
Quando falta informação, temos duas opções principais:
1. **Preencher (Imputar):** Trocar o vazio por um valor estatístico (como a média ou mediana).
2. **Excluir:** Jogar fora a linha ou a coluna inteira se faltar muita coisa.

### Exemplo: Preenchendo a Idade
Faltam várias idades no nosso dataset. Vamos preencher esses vazios com a **mediana** das idades, assim não perdemos os dados desses passageiros.

```python
# [EXEMPLO] Preenchendo valores nulos na coluna 'age' com a mediana
mediana_idade = df['age'].median()
df['age'] = df['age'].fillna(mediana_idade)

print(f"Valores nulos em 'age' após o preenchimento: {df['age'].isnull().sum()}")
```

## 2. Transformando Texto em Números (Encoding)
Modelos matemáticos não entendem palavras como "male" ou "female", eles entendem números (0 e 1). Precisamos traduzir nossas categorias.

### Exemplo: Mapeamento Simples (Dicionário)
Vamos transformar a coluna de gênero (`sex`) em uma nova coluna numérica (`sex_num`), onde homem será 0 e mulher será 1.

```python
# [EXEMPLO] Transformando texto em número com .map()
dicionario_genero = {'male': 0, 'female': 1}
df['sex_num'] = df['sex'].map(dicionario_genero)

# Vamos ver as 5 primeiras linhas dessas colunas
df[['sex', 'sex_num']].head()
```
