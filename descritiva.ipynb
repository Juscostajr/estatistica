## 1. Tratando Dados Ausentes (Imputação e Exclusão)
Quando falta informação, temos duas opções principais:
1. **Preencher (Imputar):** Trocar o vazio por um valor estatístico (como a média ou mediana).

### Exemplo: Preenchendo a Idade

## 1. Identificação de Outliers
Outliers são valores que fogem completamente do padrão dos outros dados. Uma forma clássica de encontrá-los é usando o método do **Intervalo Interquartil (IQR)**. 

### Exemplo: Encontrando os limites da Idade
Vamos calcular os limites matemáticos para descobrir quem era excepcionalmente velho ou jovem no navio.

```python
# [EXEMPLO] Calculando o IQR para a coluna 'age'
Q1_idade = df['age'].quantile(0.25)
Q3_idade = df['age'].quantile(0.75)
IQR_idade = Q3_idade - Q1_idade

limite_inferior_idade = Q1_idade - 1.5 * IQR_idade
limite_superior_idade = Q3_idade + 1.5 * IQR_idade

print(f"Limites de Idade -> Inferior: {limite_inferior_idade:.2f} | Superior: {limite_superior_idade:.2f}")

# Identificando quantos passageiros estão acima do limite superior
outliers_idade = df[df['age'] > limite_superior_idade]
print(f"Total de outliers de idade: {len(outliers_idade)}")
```
